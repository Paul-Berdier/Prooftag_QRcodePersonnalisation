```python
from pathlib import Path
import json, os
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import Image as DisplayImage, Markdown, display
RESULTS_DIR = Path(os.environ.get('E041_RESULTS_DIR','/data/e041-gamma-functional-pattern-frontier-v1'))
print(RESULTS_DIR)
```

## 1. Verdict

```python
v = json.loads((RESULTS_DIR/'verdict.json').read_text(encoding='utf-8'))
display(v)
```

## 2. Phase A — effet de gamma

```python
a = pd.read_csv(RESULTS_DIR/'phase-a-scoring/comparison.csv')
cols=[c for c in ['variant','gamma','iteration','qr_verify_exact_presets','original_exact','full_module_error_count','functional_center_error_rate','data_center_error_rate','lpips','accepted_alpha','projection_was_active','visual_guard_pass'] if c in a.columns]
display(a[cols].sort_values(['gamma','iteration']).reset_index(drop=True))
fig,ax=plt.subplots(figsize=(10,6))
for gamma,g in a.groupby('gamma'):
    g=g.sort_values('iteration'); ax.plot(g['iteration'],g['qr_verify_exact_presets'],marker='o',label=str(int(gamma)))
ax.set_xlabel('checkpoint i'); ax.set_ylabel('QR-Verify exact /37'); ax.set_title('E041 — SSR par gamma'); ax.legend(title='gamma'); plt.tight_layout(); plt.show()
```
